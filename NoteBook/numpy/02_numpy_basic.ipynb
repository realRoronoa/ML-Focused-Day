```python
"""
In real ML code, you almost never manually type out array values like np.array([1,2,3]). 
You need to generate arrays — zeros to initialize weights, random numbers to initialize neural network parameters, 
evenly spaced numbers for plotting or learning-rate schedules, identity matrices for linear algebra. 
This section is about the toolbox for generating arrays without typing every number by hand.

2. The Functions, Explained Plainly
Function	Plain English	Example
np.zeros(shape)	Array filled with all 0s	np.zeros((2,3)) → 2x3 grid of zeros
np.ones(shape)	Array filled with all 1s	np.ones((3,)) → [1,1,1]
np.full(shape, value)	Array filled with a chosen value	np.full((2,2), 7) → 2x2 grid of 7s
np.empty(shape)	Array with garbage/uninitialized values — fast because it skips filling memory. Only use when you'll overwrite every value yourself.	Unpredictable output
np.eye(n) / np.identity(n)	An n×n "identity matrix" — 1s on the diagonal, 0s elsewhere. Used constantly in linear algebra.	np.eye(3) → 3x3 with diagonal 1s
np.arange(start, stop, step)	Like Python's range(), but returns an array. Counts by fixed steps.	np.arange(0,10,2) → [0,2,4,6,8]
np.linspace(start, stop, num)	Gives you num evenly spaced numbers between start and stop (inclusive by default)	np.linspace(0,1,5) → [0, 0.25, 0.5, 0.75, 1]
np.logspace(start, stop, num)	Same idea as linspace but spaced evenly on a logarithmic (exponential) scale	Useful for learning rates, e.g. testing 0.001, 0.01, 0.1

Key difference — arange vs linspace: arange you specify the step size and it figures out how many numbers you get. linspace you specify how many numbers you want and it figures out the step size for you
"""
```

```python
'''NumPy Array Creation — Quick Notes

np.zeros((r,c)) → grid of all 0s
np.ones((r,c)) → grid of all 1s
np.full((r,c), val) → grid filled with your chosen value
np.eye(n) → identity matrix (1s on diagonal, 0s elsewhere) = "does nothing" matrix, like multiplying by 1
np.arange(start, stop, step) → like Python's range(). You give the jump size, stop is excluded.
np.linspace(start, stop, num) → you give how many numbers you want, NumPy figures out the spacing. Stop is included. Always returns floats.
np.random.rand() → random decimals, 0 to 1, flat/equal chance
np.random.randn() → random decimals, bell-curve shaped, centered at 0
np.random.randint(low, high, size) → random whole numbers
np.random.uniform(low, high, size) → random decimals, equal chance across your chosen range
np.random.normal(mean, std, size) → random decimals, bell curve, YOU pick center + spread
np.random.seed(n) → locks the "random" numbers so they repeat the same way every run. Same seed = same numbers always.

Key gotchas:

arange's count = jump-driven. linspace's count = you-decide-driven.
arange can give int output, linspace almost always gives float.
Identity matrix = matrix version of "×1" — multiply anything by it, get the same thing back.'''
```
