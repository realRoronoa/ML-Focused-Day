```python
import numpy as np
```

```python
v = np.array([3,4])

M = np.array([[1,2],[3,4]])

np.dot(v ,M)   # 32
v @ M         # 32   → the @ operator is shorthand for dot product / matrix multiply
```

```python
A = np.array([[1,2,3],[4,5,6]])   # shape (2,3)
A.T                                 # shape (3,2)
```

```python
A = np.array([[1,2],[3,4]])
print(np.linalg.det(A))   # -2.0
```

```python
A = np.array([[4,7],[2,6]])
A_inv = np.linalg.inv(A)
A @ A_inv
# [[1. 0.]
#  [0. 1.]]   ← identity matrix! confirms A_inv correctly "undoes" A
```
