```python
import numpy as np
```

```python
arr = np.array([1, 2, np.nan, 4, np.inf, 6])

np.isnan(arr)      # [False False  True False False False]
np.isinf(arr)      # [False False False False  True False]
np.isfinite(arr)   # [ True  True False  True False  True]  ← True only for "normal" real numbers

np.nan == np.nan     # False !!
```

```python
arr = np.array([1, 2, np.nan, 4, np.inf])

# Replace NaN with 0
np.nan_to_num(arr)
# [1. 2. 0. 4. 1.79e+308]   ← also replaces inf with a very large finite number

# Or manually, using boolean indexing
arr[np.isnan(arr)] = 0
```

```python
data = np.array([1, 2, np.nan, 4])
np.mean(data)       # nan  → normal mean gets "poisoned" by a single NaN
np.nanmean(data)     # 2.333... → ignores the NaN, computes mean of the rest
```

```python
'''
quick short note of this part

Part 10 — Missing/Special Values — Quick Notes

np.nan = "Not a Number" — represents missing/undefined data
np.inf = infinity — e.g. result of dividing by 0.0
Detection:
np.isnan(arr) → True where NaN
np.isinf(arr) → True where infinity
np.isfinite(arr) → True where the value is a normal, usable number (catches both NaN and Inf as "bad")
BIG GOTCHA: np.nan == np.nan is False. NaN never equals anything, not even itself. So arr == np.nan will NEVER find your NaNs — always use np.isnan() instead.
Fixing bad values:
np.nan_to_num(arr) → replaces NaN with 0, Inf with a big number, automatically
arr[np.isnan(arr)] = 0 → manual replace using boolean indexing
NaN "poisons" normal calculations:
np.mean(data) → if even ONE value is NaN, the whole result becomes NaN
np.nanmean(data) → ignores NaNs, computes using the rest → use this on real messy data
Same idea applies: nansum, nanstd, nanmax, etc.

Real use: missing sensor/survey data loads as NaN → before training any model, detect it (isnan) and handle it (fill with nanmean or drop it).
'''
```
