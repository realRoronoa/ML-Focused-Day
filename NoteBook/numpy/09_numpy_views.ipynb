```python
import numpy as np

arr = np.array([1,2,3,4,5])

# SLICING → gives a VIEW (shares memory)
sl = arr[1:4]
sl[0] = 999
print(arr)   # [1 999 3 4 5]  ← original CHANGED! Because sl was a view, not independent.
```

```python
# .copy() → explicit, guaranteed independent COPY
arr2 = np.array([1,2,3,4,5])
cp = arr2[1:4].copy()
cp[0] = 999
print(arr2)   # [1 2 3 4 5]  ← untouched, because cp is a real copy
```

```python
arr3 = np.array([1,2,3,4])
reshaped = arr3.reshape(2,2)
reshaped[0,0] = 999
print(arr3)   # [999 2 3 4]  ← changed! reshape gave a view here
```

```python
arr4 = np.array([1,2,3,4,5])
picked = arr4[[0,2,4]]     # fancy indexing
picked[0] = 999
print(arr4)   # [1 2 3 4 5]  ← unchanged, because fancy indexing always copies
```

```python
a = np.array([1,2,3,4,5])
b = a[1:4]
np.shares_memory(a, b)   # True  → b is a view

c = a[1:4].copy()
np.shares_memory(a, c)   # False → c is an independent copy
```
