```python
"""
The Types
dtype	Meaning
int8/16/32/64	Whole numbers, different byte sizes → different max ranges
float16/32/64	Decimal numbers, different precision
bool	True/False
complex64/128	Numbers with real + imaginary parts (rare in typical ML, common in signal processing)
<U21 etc.	Strings (the number = max character length NumPy reserved)

"""
```

```python
import numpy as np
```

```python
arr =  np.array([1.7,2.3,3.9])
print(arr)
print(arr.astype(int))

print(arr.astype(np.float32))


# Important quirk: np.nan == np.nan is actually False! 
# NaN is never equal to anything,
# even itself — that's why you must use np.isnan() instead of == np.nan to check for it.
```

```python
np.nan   # "Not a Number" — represents missing/undefined data
np.inf   # Infinity — e.g. result of dividing by 0.0

np.isnan(arr)     # True/False mask for NaN positions
np.isinf(arr)     # True/False mask for infinity
np.isfinite(arr)  # True where value is a normal, real number (not NaN, not inf)
```
